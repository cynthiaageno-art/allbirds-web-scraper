```python

import json
from pathlib import Path

import pandas as pd

RAW_DIR = Path("data/raw")


def main():
    files = sorted(RAW_DIR.glob("allbirds_raw_*.json"))
    dates = set()
    total_products = 0

    for file in files:
        try:
            data = json.loads(file.read_text(encoding="utf-8"))
            products = data.get("products", [])
            total_products += len(products)

            timestamp = data.get("scraped_at")
            if timestamp:
                dates.add(pd.to_datetime(timestamp).date().isoformat())

            print(
                file.name,
                "| products:", len(products),
                "| timestamp:", timestamp
            )
        except (OSError, json.JSONDecodeError) as error:
            print("Invalid file:", file.name, error)

    print("\nRaw files:", len(files))
    print("Distinct collection dates:", len(dates))
    print("Total product records across snapshots:", total_products)

    if len(dates) < 7:
        print("Seven distinct collection dates are not yet present.")
    else:
        print("Seven or more distinct dates found; verify consecutive dates.")


if __name__ == "__main__":
    main()
```
