```python

import sqlite3
from pathlib import Path

import pandas as pd

DB_PATH = Path("data/allbirds.db")


def main():
    if not DB_PATH.exists():
        raise SystemExit("Database not found. Run clean_load.py first.")

    with sqlite3.connect(DB_PATH) as conn:
        products = pd.read_sql_query(
            "SELECT * FROM products", conn
        )
        observations = pd.read_sql_query(
            "SELECT * FROM observations", conn
        )

        print("\nPRODUCT TABLE")
        print(products.head())
        print("\nOBSERVATION TABLE")
        print(observations.head())

        print("\nMissing values:")
        print(observations.isna().sum())

        print("\nDuplicate product-variant-time records:")
        print(observations.duplicated(
            ["product_id", "variant_id", "scraped_at"]
        ).sum())

        print("\nCollection timestamps:")
        print(observations["scraped_at"].nunique())

        print("\nRows by collection timestamp:")
        print(observations.groupby("scraped_at").size())

        print("\nAvailability counts:")
        print(observations["available"].value_counts(dropna=False))

        print("\nPrice summary:")
        print(observations["price"].describe())


if __name__ == "__main__":
    main()
```
